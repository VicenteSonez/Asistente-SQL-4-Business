# SQL4Business — Deliverable 2

Reproduces everything reported in `docs/deliverable2.pdf` from the repository, on the
hardware declared in Deliverable 1 (Colab T4). Nothing is uploaded by hand: the first cell
clones the repository at `REF` and installs the pinned requirements.

1. `Runtime > Change runtime type > T4 GPU`, then `Runtime > Run all`.
2. The demo runs the Deliverable 1 baseline and the structured solution on the same input.
3. The evaluation runs three systems on the 15 official questions and the 12 held-out
   paraphrases (about 25 minutes) and writes `results/deliverable2_v2.json` and
   `docs/deliverable2_results.tex`.

```python
REPO_URL = "https://github.com/VicenteSonez/Asistente-SQL-4-Business.git"
REF = "d2-correcciones"  # branch, tag or commit to evaluate

import os, subprocess, sys

ROOT = "/content/Asistente-SQL-4-Business"
if not os.path.isdir(ROOT):
    subprocess.run(["git", "clone", "--quiet", REPO_URL, ROOT], check=True)
subprocess.run(["git", "-C", ROOT, "checkout", "--quiet", REF], check=True)
subprocess.run([sys.executable, "-m", "pip", "install", "--quiet", "-r", f"{ROOT}/requirements.txt"], check=True)
os.chdir(ROOT)
print("commit:", subprocess.run(["git", "rev-parse", "HEAD"], capture_output=True, text=True).stdout.strip())
```
